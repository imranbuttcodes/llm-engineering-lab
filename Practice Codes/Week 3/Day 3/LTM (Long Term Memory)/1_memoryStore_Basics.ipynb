```python
from langgraph.store.memory import InMemoryStore
```

# Core Operations

## Every memory store supports four operations.


put()      → Save memory

get()      → Get one memory

search()   → Find memories

delete()   → Remove memory


# Every Memory Has three Things:

1. Namespace
2. Key
3. Value


think of it like 



Namespace
    users

Key
    imran

Value
    {
        "name": "Imran Butt",
        "learning": "LangGraph"
    }

# Creating Memory

```python
store = InMemoryStore()

# namespace = ("user", "u") # it means users/u1 

# store.put(namespace=namespace, key='imran', value={
#     'name': 'Imran Butt',
#     'Learning': 'LangGraph'
# })

# What happened?

# Internally the store now looks like

# users/

#     imran

#         name = Imran Butt

#         learning = LangGraph

```

```python
# creating a namespace
namespace = ("user", "u1")

# Adding memories
store.put(namespace, "1", {"data": "User likes pizza"})
store.put(namespace, "2", {"data": "User prefers dark mode"})

# another namespace
namespace2 = ("user", "u2")

# Add memories
store.put(namespace2, "1", {"data": "User likes pasta"})
store.put(namespace2, "2", {"data": "User prefers grid style navigation"})
```

# Retrieving Memories

```python
# store.get(namespace, key)
item = store.get(namespace, '1')

print(item)
print(item.namespace)
print(item.key)
print(item.value)
```

# Retrieving all Memories

```python
items = store.search(namespace)

print(items)
print()

for item in items:
    print(item)
```

# Semantic Search

```python
from langchain_huggingface import HuggingFaceEmbeddings

emb_model = HuggingFaceEmbeddings(
    model_name='sentence-transformers/all-MiniLM-L6-v2'
)
```

```python
store = InMemoryStore(index={
    'embed': emb_model,
    'dims': 1536
})
```

```python
namespace = ('users', 'u1')
```

```python
# namespace, key, value
store.put(namespace, "1", {"data": "User prefers concise answers over long explanations"})
store.put(namespace, "2", {"data": "User likes examples in Python"})
store.put(namespace, "3", {"data": "User usually works late at night"})
store.put(namespace, "4", {"data": "User prefers dark mode in applications"})
store.put(namespace, "5", {"data": "User is learning machine learning"})
store.put(namespace, "6", {"data": "User dislikes overly theoretical explanations"})
store.put(namespace, "7", {"data": "User prefers step-by-step reasoning"})
store.put(namespace, "8", {"data": "User is based in India"})
store.put(namespace, "9", {"data": "User likes real-world analogies"})
store.put(namespace, "10", {"data": "User prefers bullet points over paragraphs"})
```

```python
item = store.search(namespace, query='Which application mode do user preferes?', limit=1)
print(item)
```

```python
for item in store.search(namespace, query="What are user's preferences?", limit = 3):
    print(item.value['data'])
```
