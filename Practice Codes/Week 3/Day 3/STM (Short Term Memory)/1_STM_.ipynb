```python
from langgraph.graph import StateGraph, MessagesState, START, END
from langchain_groq import ChatGroq
from langgraph.checkpoint.memory import InMemorySaver
from dotenv import load_dotenv
import os
```

```python
load_dotenv()

llm = ChatGroq(
    model = 'llama-3.3-70b-versatile',
    api_key = os.getenv('GROQ_API_KEY')
)
```

```python
def chat_node(state: MessagesState):

    result = llm.invoke(state['messages'])

    return {
        'messages': result.content
    }
```

```python
builder = StateGraph(MessagesState)


builder.add_node('chat_node', chat_node)

builder.add_edge(START, 'chat_node')
builder.add_edge('chat_node', END)

graph = builder.compile()
```

```python
result = graph.invoke(
    {
        'messages': [
            {
                'role': 'user',
                'content': 'Hi My Name is Imran Butt'
            }
        ]
    }
)

print(result['messages'][-1].content)
```

```python
CONFIG = {
    'configurable': {
        'thread_id': 'thread_1'
    }
}
```

```python
builder = StateGraph(MessagesState)

check_pointer = InMemorySaver()

builder.add_node('chat_node', chat_node)

builder.add_edge(START, 'chat_node')
builder.add_edge('chat_node', END)

graph = builder.compile(checkpointer=check_pointer)
```

```python
result = graph.invoke(
    {
        'messages': [
            {
                'role': 'user',
                'content': 'Im imran butt'
            }
        ]
    },
    config=CONFIG
)

print(result['messages'][-1].content)
```

```python
graph.get_state(config=CONFIG)
```

```python
for state in graph.get_state(config=CONFIG):
    print(state)
```

```python
print(list(graph.get_state_history(config=CONFIG)))
```

```python
for state in graph.get_state_history(config=CONFIG):
    print(state.values)
```
