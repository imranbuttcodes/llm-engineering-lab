```python
from langchain_openai import ChatOpenAI
```

```python
reasoning_llm = ChatOpenAI(
    model = "reasoning",
    api_key = "anything",
    base_url = "http://localhost:4000"
)
```

```python
response = reasoning_llm.invoke("What is RAG in one sentence?")
```

```python
print(response.content)
```
